## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
g = 9.81

omega_0 = np.sqrt(5**2/4)*2*np.pi
def verhouding_amplitude(omega):
    return 1 / (1-omega**2/omega_0**2)

def d_x(omega_0):
    return g/omega_0**2

```

```python

x_test = np.linspace(0.001, 1000, 10000)
dx_test = np.linspace(1, 10, 100)
plt.figure()
plt.loglog(x_test, np.abs(verhouding_amplitude(x_test)))
plt.xlim(1, 1000)
plt.ylim(0.01, 1000)
plt.xlabel("omega")
plt.ylabel("verhouding amplitude")

plt.figure()
plt.plot(dx_test, d_x(dx_test))
plt.xlabel("omega_0")
plt.ylabel("delta_x")
```

```python
omega_0 = np.sqrt(5**2/4)*2*np.pi
print("omega_0 =", omega_0, "rad/s")

print("verhouding_amplitude =", 1/(1-(10*np.pi)**2/omega_0**2))

delta_x = g/(omega_0**2)

print("delta_x =", delta_x, "m")
```

## Opdracht 3: Plot de grafiek.

```python
def veerconstante(m, d_x):
    return m*g/d_x

m_test = np.linspace(0.1, 5, 1000)
plt.figure()
plt.plot(m_test, veerconstante(m_test, delta_x), label = f"delta_x = {delta_x:.2f} m")
plt.plot(m_test, veerconstante(m_test, delta_x*2), label = f"delta_x = {delta_x*2:.2f} m")
plt.plot(m_test, veerconstante(m_test, delta_x*5), label = f"delta_x = {delta_x*5:.2f} m")
plt.xlabel("massa(kg)")
plt.ylabel("veerconstante(N/m)")
plt.ylim(0,200)
plt.xlim(0, 2)
plt.legend()
```
